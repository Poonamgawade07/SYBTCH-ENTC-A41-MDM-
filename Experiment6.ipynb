```python
#empty tuple

tup=()
print(tup)
```

```python
#homogineous tuple

tup=(1,2,3,4,5)
print(tup)

tup_2=("ABC","PQR","STU")
print(tup_2)
```

```python
#heterogenious tuple

tup=("Poonam",41,9.55)
print(tup)
```

```python
#nested tuple

tup_1=("Artha",21,[7,8,9])
print(tup_1)

tup_2=("Poonam",(41,"ETC",'A'),9.5)
print(tup_2)

tup_3=("ABC",(2,4,7),[12,45,23])
print(tup_3)
```

```python
#tuple packing

tup="cat","dog","bird"
print(tup)
```
